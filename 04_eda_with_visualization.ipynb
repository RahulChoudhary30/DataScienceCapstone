# 4. EDA with Data Visualization
Charts used and why:
- **Scatter (FlightNumber vs PayloadMass, colored by Class)**: does experience and payload affect landing?
- **Scatter (FlightNumber vs LaunchSite)** and **(PayloadMass vs LaunchSite)**: site differences.
- **Bar (success rate per orbit)**: which orbits are easiest.
- **Scatter (FlightNumber vs Orbit, PayloadMass vs Orbit)**: orbit behaviour over time and weight.
- **Line (yearly success trend)**: improvement over time.

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
sns.set_theme()
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv")
df.head()
```

```python
sns.catplot(y='PayloadMass', x='FlightNumber', hue='Class', data=df, aspect=4)
plt.xlabel('Flight Number'); plt.ylabel('Pay load Mass (kg)'); plt.show()
```

```python
sns.catplot(y='LaunchSite', x='FlightNumber', hue='Class', data=df, aspect=4)
plt.xlabel('Flight Number'); plt.ylabel('Launch Site'); plt.show()
```

```python
sns.catplot(y='LaunchSite', x='PayloadMass', hue='Class', data=df, aspect=4)
plt.xlabel('Pay load Mass (kg)'); plt.ylabel('Launch Site'); plt.show()
```

## Success rate by orbit

```python
orbit_rate = df.groupby('Orbit')['Class'].mean().reset_index()
sns.barplot(x='Orbit', y='Class', data=orbit_rate)
plt.ylabel('Success rate'); plt.show()
```

```python
sns.catplot(y='Orbit', x='FlightNumber', hue='Class', data=df, aspect=4); plt.show()
sns.catplot(y='Orbit', x='PayloadMass', hue='Class', data=df, aspect=4); plt.show()
```

## Yearly trend

```python
df['Year'] = pd.to_datetime(df['Date']).dt.year
yearly = df.groupby('Year')['Class'].mean().reset_index()
sns.lineplot(x='Year', y='Class', data=yearly, marker='o')
plt.ylabel('Success rate'); plt.title('Launch success trend'); plt.show()
```

## Feature engineering
Select predictive columns, one-hot encode categoricals and save `dataset_part_3.csv` for modelling.

```python
features = df[['FlightNumber', 'PayloadMass', 'Orbit', 'LaunchSite', 'Flights', 'GridFins',
               'Reused', 'Legs', 'LandingPad', 'Block', 'ReusedCount', 'Serial']]
features_one_hot = pd.get_dummies(features, columns=['Orbit', 'LaunchSite', 'LandingPad', 'Serial'])
features_one_hot = features_one_hot.astype('float64')
features_one_hot.to_csv('dataset_part_3.csv', index=False)
features_one_hot.head()
```
