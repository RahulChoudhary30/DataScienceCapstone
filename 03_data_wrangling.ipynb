# 3. Data Wrangling
**Goal:** explore launch sites, orbits and landing outcomes, then create the binary label `Class` (1 = first stage landed, 0 = did not).

```python
import pandas as pd, numpy as np
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv")
df.head()
```

```python
df.isnull().sum() / len(df) * 100
```

## Launches per site, per orbit and mission outcomes

```python
print(df['LaunchSite'].value_counts(), '\n')
print(df['Orbit'].value_counts(), '\n')
landing_outcomes = df['Outcome'].value_counts()
landing_outcomes
```

```python
for i, outcome in enumerate(landing_outcomes.keys()):
    print(i, outcome)
```

## Create the landing class
Outcomes `False ASDS`, `False Ocean`, `False RTLS`, `None ASDS`, `None None` are unsuccessful (0); everything else is a successful landing (1).

```python
bad_outcomes = set(landing_outcomes.keys()[[1, 3, 5, 6, 7]])
print(bad_outcomes)
df['Class'] = df['Outcome'].apply(lambda x: 0 if x in bad_outcomes else 1)
df[['Class']].head(8)
```

```python
print('Overall success rate:', df['Class'].mean())
df.to_csv('dataset_part_2.csv', index=False)
```
